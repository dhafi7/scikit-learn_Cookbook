## 1. Pendahuluan
Dalam banyak aplikasi (seperti keamanan siber atau deteksi penipuan), kita ingin mendeteksi data yang sangat menyimpang dari pola normal. `scikit-learn` secara konvensi memberikan label **1 untuk data inliers (normal)** dan **-1 untuk data outliers (anomali)**.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.neighbors import LocalOutlierFactor

%matplotlib inline
np.random.seed(42)
```

## 2. Isolation Forest (Outlier Detection)
Sangat cepat dan efisien. Algoritma ini mencoba mengisolasi setiap titik data secara acak. Outlier adalah titik-titik yang paling cepat terisolasi (membutuhkan sedikit belahan/split).

```python
# Simulasi Data Normal
X_normal = 0.3 * np.random.randn(200, 2)
# Simulasi Data Outlier
X_outliers = np.random.uniform(low=-4, high=4, size=(20, 2))
X_train = np.r_[X_normal, X_outliers]

# Melatih Isolation Forest (asumsi 10% data adalah kontaminasi/outlier)
iso_forest = IsolationForest(contamination=0.1, random_state=42)
y_pred_iso = iso_forest.fit_predict(X_train)

plt.figure(figsize=(7, 5))
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_pred_iso, cmap='coolwarm', edgecolor='k')
plt.title('Isolation Forest (Merah/1: Normal, Biru/-1: Outlier)')
plt.show()
```

## 3. One-Class SVM (Novelty Detection)
Digunakan jika kita memiliki data training yang 100% bersih dari anomali. Model akan membentuk perbatasan ketat, lalu mendeteksi apakah data uji baru (*new data*) berada di luar perbatasan tersebut.

```python
# Dilatih HANYA pada data normal (bersih)
oc_svm = OneClassSVM(kernel='rbf', gamma='auto', nu=0.05)
oc_svm.fit(X_normal)

# Data Test Baru (Campuran)
X_test_new = np.r_[0.3 * np.random.randn(50, 2), np.random.uniform(low=-4, high=4, size=(10, 2))]
y_pred_svm = oc_svm.predict(X_test_new)

plt.figure(figsize=(7, 5))
plt.scatter(X_test_new[:, 0], X_test_new[:, 1], c=y_pred_svm, cmap='Set1', edgecolor='k')
plt.title('One-Class SVM Mendeteksi Data Baru (Merah: Outlier, Abu: Normal)')
plt.show()
```
