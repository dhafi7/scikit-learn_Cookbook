## 1. Pendahuluan
Decision Tree bekerja dengan membelah ruang data menggunakan aturan pertanyaan bersyarat (*if-else*). Kita akan mengeksplorasi dari model pohon tunggal hingga metode *ensemble* tingkat lanjut.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

%matplotlib inline
np.random.seed(42)

X, y = make_classification(n_samples=500, n_features=4, n_informative=2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 2. Decision Tree Classifier
Sangat interpretabel, namun pohon yang terlalu dalam rentan menghafal data latih (*overfitting*). Kita akan membatasi kedalamannya (`max_depth`).

```python
dt = DecisionTreeClassifier(max_depth=3, random_state=42)
dt.fit(X_train, y_train)
print(f"Akurasi Decision Tree: {accuracy_score(y_test, dt.predict(X_test)):.3f}")

plt.figure(figsize=(12, 6))
plot_tree(dt, filled=True, rounded=True, class_names=['0', '1'])
plt.title('Visualisasi Decision Tree (max_depth=3)')
plt.show()
```

## 3. Ensemble: Random Forest & Gradient Boosting
Metode *ensemble* menggabungkan ratusan pohon keputusan menjadi satu prediktor super.

```python
# Random Forest (Paralel)
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
print(f"Akurasi Random Forest: {accuracy_score(y_test, rf.predict(X_test)):.3f}")

# Gradient Boosting (Sekuensial)
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gb.fit(X_train, y_train)
print(f"Akurasi Gradient Boosting: {accuracy_score(y_test, gb.predict(X_test)):.3f}")

# Fitur Importance
plt.figure(figsize=(6, 4))
plt.bar([f'Fitur {i}' for i in range(4)], gb.feature_importances_, color='teal')
plt.title('Fitur Importance dari Gradient Boosting')
plt.show()
```
