## 1. Pendahuluan
Mendapatkan akurasi tinggi pada *train_test_split* tunggal bisa jadi hanya kebetulan. Kita membutuhkan Cross-Validation untuk memastikan konsistensi, dan SearchCV untuk mencari parameter terbaik secara sistematis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, GridSearchCV, RandomizedSearchCV, StratifiedKFold
from scipy.stats import randint

%matplotlib inline
np.random.seed(42)

data = load_breast_cancer()
X, y = data.data, data.target
```

## 2. Stratified K-Fold Cross-Validation
Untuk masalah klasifikasi, kita wajib menggunakan `StratifiedKFold` agar setiap lipatan pengujian mempertahankan keseimbangan kelas yang sama (mencegah ada lipatan yang isinya hanya 1 kelas saja).

```python
rf = RandomForestClassifier(n_estimators=50, random_state=42)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(rf, X, y, cv=skf, scoring='accuracy')

print(f"Akurasi di tiap lipatan (5-fold): {scores}")
print(f"Rata-rata Akurasi: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

## 3. Randomized Search CV
Mencari parameter menggunakan Grid Search pada semua kemungkinan kombinasi akan sangat lama. `RandomizedSearchCV` mencoba sampel acak dari distribusi parameter yang kita berikan, jauh lebih cepat dengan hasil yang hampir sama optimalnya.

```python
# Distribusi parameter yang akan dicari
param_dist = {
    'n_estimators': randint(20, 200),
    'max_depth': randint(2, 20),
    'min_samples_split': randint(2, 11)
}

# Mencoba 10 kombinasi acak (n_iter=10)
random_search = RandomizedSearchCV(rf, param_distributions=param_dist, n_iter=10, cv=3, random_state=42)
random_search.fit(X, y)

print("Parameter Terbaik yang Ditemukan:")
print(random_search.best_params_)
print(f"Akurasi Cross-Validation Terbaik: {random_search.best_score_:.4f}")
```
