## 1. Introduction
Achieving high accuracy on a single *train_test_split* could simply be a matter of luck. We require Cross-Validation to verify performance consistency, and SearchCV methods to systematically explore and pinpoint the best model parameters.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, GridSearchCV, RandomizedSearchCV, StratifiedKFold
from scipy.stats import randint

%matplotlib inline
np.random.seed(42)

data = load_breast_cancer()
X, y = data.data, data.target
```

## 2. Stratified K-Fold Cross-Validation
For classification tasks, it is crucial to employ `StratifiedKFold`. This ensures that each verification fold preserves the exact same class balance distribution as the overall dataset, preventing scenarios where a fold contains only a single target class.

```python
rf = RandomForestClassifier(n_estimators=50, random_state=42)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(rf, X, y, cv=skf, scoring='accuracy')

print(f"Accuracy in each fold (5-fold): {scores}")
print(f"Mean Accuracy: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

## 3. Randomized Search CV
Evaluating parameters using Grid Search across every single combination can be extremely slow. `RandomizedSearchCV` samples a fixed number of configurations from specified parameter distributions at random, executing much faster while delivering comparably optimal results.

```python
# Parameter distributions to explore
param_dist = {
    'n_estimators': randint(20, 200),
    'max_depth': randint(2, 20),
    'min_samples_split': randint(2, 11)
}

# Evaluate 10 random combinations (n_iter=10)
random_search = RandomizedSearchCV(rf, param_distributions=param_dist, n_iter=10, cv=3, random_state=42)
random_search.fit(X, y)

print("Best Parameters Discovered:")
print(random_search.best_params_)
print(f"Best Cross-Validation Accuracy Score: {random_search.best_score_:.4f}")
```
