## 1. Pendahuluan
Ekstraksi fitur dari teks adalah tantangan tersendiri. Bab ini mendemonstrasikan bagaimana kita membersihkan teks, melakukan vektorisasi (mengubah teks menjadi representasi matriks angka), dan menggunakannya untuk klasifikasi multi-kelas.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report

# Simulasi data teks: 3 kategori (Teknologi, Olahraga, Makanan)
corpus = [
    "Apple rilis smartphone baru dengan prosesor AI canggih",
    "Google mengupdate algoritma machine learning di cloud",
    "Timnas menang dramatis lewat gol menit akhir di final piala AFF",
    "Pemain basket itu mencetak 50 poin dalam satu pertandingan",
    "Resep membuat nasi goreng jawa yang enak dan lezat",
    "Restoran ramen baru ini menyajikan kaldu ayam yang gurih"
]
y_labels = ['Teknologi', 'Teknologi', 'Olahraga', 'Olahraga', 'Makanan', 'Makanan']

print("Dataset Teks Asli:")
for i, text in enumerate(corpus):
    print(f"- {y_labels[i]}: '{text}'")
```

## 2. Vektorisasi Teks (TF-IDF)
Kita menggunakan TF-IDF. Ia tidak hanya menghitung kemunculan kata, tetapi juga menurunkankan bobot kata yang muncul terlalu sering di semua dokumen.

```python
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(corpus)

print(f"Ukuran Matriks TF-IDF: {X_tfidf.shape} (Ada {X_tfidf.shape[1]} kata unik/fitur)")
print("Beberapa fitur kata yang diekstrak:", vectorizer.get_feature_names_out()[:10])
```

## 3. Klasifikasi Multiclass dengan Pipeline
Kita gunakan algoritma Naive Bayes tipe `MultinomialNB` yang sangat cocok untuk representasi teks/frekuensi. Untuk rapi, kita satukan ekstraksi dan prediksi dalam sebuah `Pipeline`.

```python
# Membuat Pipeline text-to-prediction
model_pipeline = make_pipeline(TfidfVectorizer(), MultinomialNB())

# Melatih model
model_pipeline.fit(corpus, y_labels)

# Data Uji Baru (Data yang tidak pernah dilihat model)
test_docs = [
    "Cara memasak mie ayam super pedas",
    "Microsoft mengembangkan software AI untuk laptop",
    "Pelatih sepakbola tersebut merubah taktik babak kedua"
]
y_true = ['Makanan', 'Teknologi', 'Olahraga']

# Prediksi
y_pred = model_pipeline.predict(test_docs)

print("=== Hasil Prediksi Data Baru ===")
for doc, pred in zip(test_docs, y_pred):
    print(f"Teks: '{doc}' \n--> Prediksi: {pred}\n")
    
print("Classification Report:")
print(classification_report(y_true, y_pred))
```
