## 1. Pendahuluan
Clustering adalah teknik Unsupervised Learning. Di bab ini, kita membandingkan algoritma K-Means tradisional dengan algoritma yang lebih canggih seperti MiniBatch K-Means dan DBSCAN yang mampu menangani data dengan bentuk acak.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, make_blobs
from sklearn.cluster import KMeans, MiniBatchKMeans, DBSCAN
from sklearn.metrics import silhouette_score
import time

%matplotlib inline
np.random.seed(42)
```

## 2. K-Means vs MiniBatch K-Means
MiniBatch K-Means menggunakan sebagian kecil dari data pada setiap iterasinya, menjadikannya pilihan utama untuk dataset berskala besar.

```python
# Dataset dengan 10.000 titik
X_large, y_large = make_blobs(n_samples=10000, centers=5, random_state=42)

start_time = time.time()
kmeans = KMeans(n_clusters=5, n_init=10, random_state=42)
kmeans.fit(X_large)
print(f"Waktu komputasi K-Means standar: {time.time() - start_time:.3f} detik")

start_time = time.time()
mb_kmeans = MiniBatchKMeans(n_clusters=5, batch_size=100, n_init=10, random_state=42)
mb_kmeans.fit(X_large)
print(f"Waktu komputasi MiniBatch K-Means: {time.time() - start_time:.3f} detik (Jauh lebih cepat!)")
```

## 3. Menangani Data Non-Spherikal dengan DBSCAN
K-Means gagal pada dataset melengkung atau memanjang. DBSCAN mendeteksi area dengan kepadatan tinggi dan bisa membentuk klaster dengan rupa apapun.

```python
# Dataset bentuk bulan sabit
X_moons, _ = make_moons(n_samples=500, noise=0.05, random_state=42)

# 1. K-Means (Akan gagal)
km_labels = KMeans(n_clusters=2, n_init=10, random_state=42).fit_predict(X_moons)

# 2. DBSCAN (Akan berhasil)
# eps = radius pencarian tetangga, min_samples = minimal titik untuk disebut 'kepadatan tinggi'
dbscan_labels = DBSCAN(eps=0.2, min_samples=5).fit_predict(X_moons)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].scatter(X_moons[:, 0], X_moons[:, 1], c=km_labels, cmap='viridis')
ax[0].set_title('Kegagalan K-Means pada Data Bulan Sabit')

ax[1].scatter(X_moons[:, 0], X_moons[:, 1], c=dbscan_labels, cmap='viridis')
ax[1].set_title('Keberhasilan DBSCAN (Berbasis Kepadatan)')
plt.show()
```

## 4. Evaluasi dengan Silhouette Score
Rentang skor adalah -1 hingga 1. Mendekati 1 berarti klaster sangat terpisah dan padat. Mendekati 0 berarti klaster tumpang tindih.

```python
score = silhouette_score(X_large, kmeans.labels_)
print(f"Silhouette Score untuk K-Means pada dataset X_large: {score:.3f}")
```
